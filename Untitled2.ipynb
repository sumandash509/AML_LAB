```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Target Variable
y_name = input("Enter Target Variable Name: ")
y = np.array(list(map(float, input(f"Enter {y_name} values (comma separated): ").split(','))))

# Number of Features
n = int(input("Enter Number of Features: "))

data = {}
feature_names = []

for i in range(n):
    name = input(f"Enter Feature {i+1} Name: ")
    feature_names.append(name)

    values = np.array(list(map(float, input(f"Enter {name} values: ").split(','))))

    if len(values) != len(y):
        print("Error: Number of values must be same.")
        exit()

    data[name] = values

X = pd.DataFrame(data)

# Model Training
model = LinearRegression()
model.fit(X, y)

pred = model.predict(X)

# Metrics
mse = mean_squared_error(y, pred)
mae = mean_absolute_error(y, pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, pred)

# Regression Equation
print("\n------ Regression Equation ------")
equation = f"{y_name} = {model.intercept_:.2f}"

for coef, name in zip(model.coef_, feature_names):
    equation += f" + ({coef:.2f})*{name}"

print(equation)

print("\nIntercept =", round(model.intercept_,2))

for coef, name in zip(model.coef_, feature_names):
    print(f"Coefficient of {name} =", round(coef,2))

print("\nMSE  =", round(mse,4))
print("MAE  =", round(mae,4))
print("RMSE =", round(rmse,4))
print("R2 Score =", round(r2,4))

# Prediction
choice = input("\nDo you want Prediction? (yes/no): ").lower()

if choice == "yes":
    sample = []

    for name in feature_names:
        sample.append(float(input(f"Enter {name}: ")))

    sample_df = pd.DataFrame([sample], columns=feature_names)

    result = model.predict(sample_df)

    print(f"\nPredicted {y_name} =", round(result[0],2))

# Scatter Plot
plt.figure(figsize=(6,5))
plt.scatter(y, pred, color='blue')
plt.plot([min(y), max(y)], [min(y), max(y)], 'r--')
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs Predicted")
plt.grid(True)
plt.show()

# Performance Chart
metrics = ["R2", "RMSE", "MAE", "MSE"]
values = [r2, rmse, mae, mse]

plt.figure(figsize=(6,4))
bars = plt.bar(metrics, values)

for bar in bars:
    plt.text(bar.get_x()+0.15, bar.get_height(), round(bar.get_height(),2))

plt.title("Linear Regression Performance")
plt.ylabel("Value")
plt.grid(axis="y")
plt.show()
```
