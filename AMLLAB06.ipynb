```python
# Chi-Square Test of Independence

tab = float(input("Enter the tabulated Chi-square value: "))
r = int(input("Enter number of rows: "))

obs = []
for i in range(r):
    obs.append(list(map(float, input(f"Row {i+1}: ").split())))

c = len(obs[0])
rt = [sum(row) for row in obs]
ct = [sum(obs[i][j] for i in range(r)) for j in range(c)]
gt = sum(rt)

chi = 0

print("\nRow Col | Observed | Expected | (O-E)^2/E")

for i in range(r):
    for j in range(c):
        o = obs[i][j]
        e = rt[i] * ct[j] / gt
        x = (o - e) ** 2 / e
        chi += x
        print(i+1, j+1, f"{o:.2f} {e:.4f} {x:.4f}")

df = (r-1) * (c-1)

print("\n--- Summary ---")
print("Degrees of Freedom (df):", df)
print(f"Calculated Chi-square value: {chi:.4f}")
print(f"Tabulated Chi-square value: {tab:.4f}")

if chi < tab:
    print("Decision: ACCEPT the Null Hypothesis (H0)")
else:
    print("Decision: REJECT the Null Hypothesis (H0)")
```
